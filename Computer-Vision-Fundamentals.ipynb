```python
import cv2
```

```python
print(cv2.__version__)
```

```python
img=cv2.imread("bird.jpg",0)
```

```python
cv2.imshow("bird image" , img)  
k= cv2.waitKey(0)
if k == 27:
    cv2.destroyAllWindows()
elif k== ord('s'):
    cv2.imwrite("gray bird.jpg" , img)    
    cv2.destroyAllWindows()



```
