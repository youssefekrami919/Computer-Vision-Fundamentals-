# Image Reading 

```python
import cv2
```

```python
print(cv2.__version__)
```

```python
img=cv2.imread("bird.jpg",0)
```

```python
cv2.imshow("bird image" , img)  
k= cv2.waitKey(0)
if k == 27:
    cv2.destroyAllWindows()
elif k== ord('s'):
    cv2.imwrite("gray bird.jpg" , img)    
    cv2.destroyAllWindows()



```

# Vedio Capture

```python
capture= cv2.VideoCapture(0)
fourcc=cv2.VideoWriter_fourcc(*'XVID')
output= cv2.VideoWriter("capture1.mp4", fourcc,20 , (640,480))


print(capture.isOpened())
while (capture.isOpened()) :
    ret ,frame=capture.read()
    if ret== True:
        output.write(frame)

```
