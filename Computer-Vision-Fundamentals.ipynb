# Vedio Capture

```python
import cv2

capture = cv2.VideoCapture(0)

# قراءة أبعاد الكاميرا الحقيقية لتجنب تلف الفيديو
width = int(capture.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(capture.get(cv2.CAP_PROP_FRAME_HEIGHT))

# استخدام mp4v مع صيغة mp4
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
output = cv2.VideoWriter('capture1.mp4', fourcc, 20.0, (width, height))

while capture.isOpened():
  ret, frame = capture.read()

  if not ret:
    print('Fail to read the camera')
    break

  # حفظ الفريم وعرضه
  output.write(frame)
  cv2.imshow('Video Capture', frame)

  # الخروج عند الضغط على زر 'q'
  if cv2.waitKey(1) == ord('q'):
    break

capture.release()
output.release()
cv2.destroyAllWindows()
```

```python
from cv2 import VideoCapture
import cv2
import datetime


font = cv2.FONT_HERSHEY_SIMPLEX

cap = cv2.VideoCapture(0)
width= int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height= int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

print("width value : " , width)
print("height value : " , height)



while cap.isOpened():
    ret , frame = cap.read()
    if ret == True:
        text = "width : "+ str(cap.get(3)) + " hight : "+str(cap.get(4))
        date=str(datetime.datetime.now())
        frame=cv2.putText(frame , text , (10,50),font, 1,(0,255,255),2)
        frame=cv2.putText(frame , date , (10,100),font, 1,(0,255,255),2)
  
        cv2.imshow("Date & Time showing " , frame)
        if cv2.waitKey(1)== ord('q'):
            break
    else:
        break        

cap.release()
cv2.destroyAllWindows()

    



```

```python
import cv2
from cv2 import waitKey

def click_event(event , x , y , flags , param):
    if event== cv2.EVENT_LBUTTONDOWN:
        print(x,",",y)
        font=cv2.FONT_HERSHEY_SIMPLEX
        strxy=str(x) + "," + str(y)
        cv2.putText(img , strxy , (x,y) , font ,.5 , (255,255,0), 2)
        cv2.imshow("strawberry" , img)




cv2.imshow("strawberry" , img)
cv2.setMouseCallback("strawberry", click_event) 

k=waitKey(0)
if k==27:
    cv2.destroyAllWindows()

cv2.setMouseCallback("strawberry.jpg")        
        
```
