# Vedio Capture

```python
import cv2

capture = cv2.VideoCapture(0)

# قراءة أبعاد الكاميرا الحقيقية لتجنب تلف الفيديو
width = int(capture.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(capture.get(cv2.CAP_PROP_FRAME_HEIGHT))

# استخدام mp4v مع صيغة mp4
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
output = cv2.VideoWriter('capture1.mp4', fourcc, 20.0, (width, height))

while capture.isOpened():
  ret, frame = capture.read()

  if not ret:
    print('Fail to read the camera')
    break

  # حفظ الفريم وعرضه
  output.write(frame)
  cv2.imshow('Video Capture', frame)

  # الخروج عند الضغط على زر 'q'
  if cv2.waitKey(1) == ord('q'):
    break

capture.release()
output.release()
cv2.destroyAllWindows()
```

## Bind Trackbar To OpenCV Window

```python
import cv2 as cv
import numpy as np

def nothing(x):
    print(x)

img = np.zeros((300,512,3), np.uint8)
cv.namedWindow('image')

cv.createTrackbar('B', 'image', 0, 255, nothing)
cv.createTrackbar('G', 'image', 0, 255, nothing)
cv.createTrackbar('R', 'image', 0, 255, nothing)

switch = '0 : OFF\n 1 : ON'
cv.createTrackbar(switch, 'image', 0, 1, nothing)

while(1):
    cv.imshow('image',img)    
    k = cv.waitKey(1) 
    if k == 27:#esc    
        break        

    b = cv.getTrackbarPos('B', 'image')
    g = cv.getTrackbarPos('G', 'image')
    r = cv.getTrackbarPos('R', 'image')
    s = cv.getTrackbarPos(switch, 'image')

    if s == 0:
       img[:] = 0   
    else:
       img[:] = [b, g, r]   


cv.destroyAllWindows()

```

# Auto color Recognition 

```python
import cv2

cap = cv2.VideoCapture(0)
cap.set(cv2.CAP_PROP_FRAME_WIDTH, 1280)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 720)

while True:
    _, frame = cap.read()
    hsv_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)
    height, width, _ = frame.shape

    cx = int(width / 2)
    cy = int(height / 2)

    # Pick pixel value
    pixel_center = hsv_frame[cy, cx]
    hue_value = pixel_center[0]

    color = "Undefined"
    if hue_value < 5:
        color = "RED"
    elif hue_value < 22:
        color = "ORANGE"
    elif hue_value < 33:
        color = "YELLOW"
    elif hue_value < 78:
        color = "GREEN"
    elif hue_value < 131:
        color = "BLUE"
    elif hue_value < 170:
        color = "VIOLET"
    else:
        color = "RED"

    pixel_center_bgr = frame[cy, cx]
    b, g, r = int(pixel_center_bgr[0]), int(pixel_center_bgr[1]), int(pixel_center_bgr[2])

    cv2.rectangle(frame, (cx - 220, 10), (cx + 200, 120), (255, 255, 255), -1)
    cv2.putText(frame, color, (cx - 200, 100), 0, 3, (b, g, r), 5)
    cv2.circle(frame, (cx, cy), 5, (25, 25, 25), 3)
    cv2.imshow("Frame", frame)
    #out.write(frame) #save your video
    key = cv2.waitKey(1)
    if key == 27:
        break

cap.release()

cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

# 1. حساب الطاقة (أماكن الحواف والتفاصيل في الصورة)
def get_energy(img):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    sobel_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0)
    sobel_y = cv2.Sobel(gray, cv2.CV_64F, 0, 1)
    return np.abs(sobel_x) + np.abs(sobel_y)

# 2. إيجاد المسار الأقل أهمية من الأعلى للأسفل
def find_seam(energy):
    h, w = energy.shape
    cost = energy.copy()
    
    # تجميع أقل طاقة لكل بكسل من الصف السابق
    for i in range(1, h):
        for j in range(w):
            left = cost[i - 1, j - 1] if j > 0 else float("inf")
            mid  = cost[i - 1, j]
            right = cost[i - 1, j + 1] if j < w - 1 else float("inf")
            cost[i, j] += min(left, mid, right)
            
    # تتبع المسار الأقل رجوعاً للأعلى (Backtracking)
    seam = [np.argmin(cost[-1])]
    for i in range(h - 2, -1, -1):
        prev_col = seam[-1]
        c_left = cost[i, prev_col - 1] if prev_col > 0 else float("inf")
        c_mid  = cost[i, prev_col]
        c_right = cost[i, prev_col + 1] if prev_col < w - 1 else float("inf")
        
        # اختيار أصغر عمود مجاور
        offset = np.argmin([c_left, c_mid, c_right]) - 1
        seam.append(prev_col + offset)
        
    seam.reverse() # نعكس الترتيب ليبدأ من الصف 0 إلى الأخير
    return seam

# 3. حذف المسار وتقليص العرض بـ 1 بكسل
def remove_seam(img, seam):
    h, w, c = img.shape
    new_img = np.zeros((h, w - 1, c), dtype=img.dtype)
    for i in range(h):
        col = seam[i]
        # دمج كل البكسلات قبل وبعد هذا العمود
        new_img[i] = np.delete(img[i], col, axis=0)
    return new_img

# =========================
# الاستخدام المباشر
# =========================
img = cv2.imread("strawberry.jpg") # ضع اسم صورتك هنا

# عدد البكسلات المراد حذفها من العرض
num_pixels_to_remove = 30 

for _ in range(num_pixels_to_remove):
    energy = get_energy(img)
    seam = find_seam(energy)
    img = remove_seam(img, seam)

cv2.imwrite("output_simple.jpg", img)
print("تم تصغير الصورة بنجاح!")
```
